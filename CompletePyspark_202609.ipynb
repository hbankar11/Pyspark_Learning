```python
# Defining schema using nested StructType
from pyspark.sql.types import StructType,StructField,StringType,IntegerType
structureData = [
    (("James","","Smith"),"36636","M",3100),
    (("Michael","Rose",""),"40288","M",4300),
    (("Robert","","Williams"),"42114","M",1400),
    (("Maria","Anne","Jones"),"39192","F",5500),
    (("Jen","Mary","Brown"),"","F",-1)
]

nested_schema = StructType([
                  StructField("name",StructType([
                                        StructField("firstname",StringType(),True),
                                        StructField("middlename",StringType(),True),
                                        StructField("lastname",StringType(),True),]),True),
                  StructField("id",StringType(),True),
                  StructField("gender",StringType(),True),
                  StructField("salary",IntegerType(),True),
])

df = spark.createDataFrame(structureData,nested_schema)
df.show(truncate=False)
'''
+--------------------+-----+------+------+
|name                |id   |gender|salary|
+--------------------+-----+------+------+
|{James, , Smith}    |36636|M     |3100  |
|{Michael, Rose, }   |40288|M     |4300  |
|{Robert, , Williams}|42114|M     |1400  |
|{Maria, Anne, Jones}|39192|F     |5500  |
|{Jen, Mary, Brown}  |     |F     |-1    |
+--------------------+-----+------+------+
'''


```

```python
# Adding & Changing struct of the DataFrame:
# struct() - using this we can add new struct type to it.

from pyspark.sql.functions import col,struct,when

updated_df = df.withColumn("otherInfo",
                           struct(col("id").alias("identifier"),
                                  col("gender").alias("gender"),
                                  col("salary").alias("salary"),
                                  when(col("salary").cast(IntegerType()) < 2000,"Low")
                                  .when(col("salary").cast(IntegerType()) < 4000, "Medium")
                                  .otherwise("High").alias("Salary_Grade"))).drop("id","gender","salary")
updated_df.printSchema()
updated_df.show(truncate=False)
'''
root
 |-- name: struct (nullable = true)
 |    |-- firstname: string (nullable = true)
 |    |-- middlename: string (nullable = true)
 |    |-- lastname: string (nullable = true)
 |-- otherInfo: struct (nullable = false)
 |    |-- identifier: string (nullable = true)
 |    |-- gender: string (nullable = true)
 |    |-- salary: integer (nullable = true)
 |    |-- Salary_Grade: string (nullable = false)

+--------------------+------------------------+
|name                |otherInfo               |
+--------------------+------------------------+
|{James, , Smith}    |{36636, M, 3100, Medium}|
|{Michael, Rose, }   |{40288, M, 4300, High}  |
|{Robert, , Williams}|{42114, M, 1400, Low}   |
|{Maria, Anne, Jones}|{39192, F, 5500, High}  |
|{Jen, Mary, Brown}  |{, F, -1, Low}          |
+--------------------+------------------------+

'''

```

```python
#dropFields() - method returns a new PySpark Column object with the specified nested fields removed
from pyspark.sql import Row

data = [Row(name="Alex", age=20, friend=Row(name="Bob",age=30,height=150)),
        Row(name="Cathy", age=40, friend=Row(name="Doge",age=40,height=180))]
df = spark.createDataFrame(data)
df.printSchema()
'''
root
 |-- name: string (nullable = true)
 |-- age: long (nullable = true)
 |-- friend: struct (nullable = true)
 |    |-- name: string (nullable = true)
 |    |-- age: long (nullable = true)
 |    |-- height: long (nullable = true)
 '''
 # Here we need to remove nested fields like age and height
updated_col = df['friend'].dropFields("age","height")
df_new = df.withColumn("friend",updated_col)
df_new.printSchema()
'''
root
 |-- name: string (nullable = true)
 |-- age: long (nullable = true)
 |-- friend: struct (nullable = true)
 |    |-- name: string (nullable = true)

'''
```

```python
# withField () - used to either add or update a nested field value
from pyspark.sql import Row
data = [
    Row(name="Alex", age=20, friend=Row(name="Bob",age=30)),
    Row(name="Cathy", age=40, friend=Row(name="Doge",age=40))
]
df = spark.createDataFrame(data)
df.printSchema()
'''
root
 |-- name: string (nullable = true)
 |-- age: long (nullable = true)
 |-- friend: struct (nullable = true)
 |    |-- name: string (nullable = true)
 |    |-- age: long (nullable = true)
'''

#Updating value in nested column in feild:
import pyspark.sql.functions as f
updated_col = df["friend"].withField("name",f.lit("Bob"))
df.withColumn("friend",updated_col).show()
'''
+-----+---+---------+
| name|age|   friend|
+-----+---+---------+
| Alex| 20|{Bob, 30}|
|Cathy| 40|{Bob, 40}|
+-----+---+---------+
'''

#Adding new field values in nested rows in pyspark
updated_col = df["friend"].withField("upper_name", f.upper("friend.name"))
df_new = df.withColumn("friend", updated_col)
df_new.show()
'''
+-----+---+----------------+
| name|age|          friend|
+-----+---+----------------+
| Alex| 20|  {Bob, 30, BOB}|
|Cathy| 40|{Doge, 40, DOGE}|
+-----+---+----------------+
'''
```

```python
# PySpark - Filter():
from pyspark.sql.types import StructType,StructField,StringType,ArrayType

data =[
(("James","","Smith"),["Java","Scala","C++"],"OH","M"),
(("Anna","Rose",""),["Spark","Java","C++"],"NY","F"),
(("Julia","","Williams"),["CSharp","VB"],"OH","F"),
(("Maria","Anne","Jones"),["CSharp","VB"],"NY","M"),
(("Jen","Mary","Brown"),["CSharp","VB"],"NY","M"),
(("Mike","Mary","Williams"),["Python","VB"],"OH","M")
]

schema = StructType([
     StructField('name', StructType([
        StructField('firstname',StringType(),True),
        StructField('middlename',StringType(),True),
         StructField('lastname',StringType(),True)
])),
     StructField('languages',ArrayType(StringType()),True),
     StructField('state',StringType(),True),
     StructField('gender',StringType(),True)
])

df = spark.createDataFrame(data,schema)
df.printSchema()
'''
root
 |-- name: struct (nullable = true)
 |    |-- firstname: string (nullable = true)
 |    |-- middlename: string (nullable = true)
 |    |-- lastname: string (nullable = true)
 |-- languages: array (nullable = true)
 |    |-- element: string (containsNull = true)
 |-- state: string (nullable = true)
 |-- gender: string (nullable = true)
'''
df.show(truncate=False)
'''
+----------------------+------------------+-----+------+
|name                  |languages         |state|gender|
+----------------------+------------------+-----+------+
|{James, , Smith}      |[Java, Scala, C++]|OH   |M     |
|{Anna, Rose, }        |[Spark, Java, C++]|NY   |F     |
|{Julia, , Williams}   |[CSharp, VB]      |OH   |F     |
|{Maria, Anne, Jones}  |[CSharp, VB]      |NY   |M     |
|{Jen, Mary, Brown}    |[CSharp, VB]      |NY   |M     |
|{Mike, Mary, Williams}|[Python, VB]      |OH   |M     |
+----------------------+------------------+-----+------+
'''

# 1: DataFrame filter() with Column Condition
from pyspark.sql.functions import col
df.filter(col("state") == 'OH').show(truncate=False)
df.filter("state == 'OH'").show(truncate=False)
'''
+----------------------+------------------+-----+------+
|name                  |languages         |state|gender|
+----------------------+------------------+-----+------+
|{James, , Smith}      |[Java, Scala, C++]|OH   |M     |
|{Julia, , Williams}   |[CSharp, VB]      |OH   |F     |
|{Mike, Mary, Williams}|[Python, VB]      |OH   |M     |
+----------------------+------------------+-----+------+
'''

#2: Using not equal filter condition
df.filter(col("state") != 'OH').show(truncate=False)
df.filter(~(col("state") == 'OH')).show(truncate=False)
df.filter("state <> 'OH'").show(truncate=False)

'''
+--------------------+------------------+-----+------+
|name                |languages         |state|gender|
+--------------------+------------------+-----+------+
|{Anna, Rose, }      |[Spark, Java, C++]|NY   |F     |
|{Maria, Anne, Jones}|[CSharp, VB]      |NY   |M     |
|{Jen, Mary, Brown}  |[CSharp, VB]      |NY   |M     |
+--------------------+------------------+-----+------+
'''

#4: Filtering with SQL Expression
df.filter("gender == 'M'").show(truncate=False)
df.filter("gender != 'M'").show(truncate=False)
df.filter("gender <> 'M'").show(truncate=False)

'''
+-------------------+------------------+-----+------+
|name               |languages         |state|gender|
+-------------------+------------------+-----+------+
|{Anna, Rose, }     |[Spark, Java, C++]|NY   |F     |
|{Julia, , Williams}|[CSharp, VB]      |OH   |F     |
+-------------------+------------------+-----+------+
'''

#5. PySpark Filter with Multiple Conditions
df.filter((df.state == 'OH') & (df.gender == 'M')).show(truncate=False)
'''
+----------------------+------------------+-----+------+
|name                  |languages         |state|gender|
+----------------------+------------------+-----+------+
|{James, , Smith}      |[Java, Scala, C++]|OH   |M     |
|{Mike, Mary, Williams}|[Python, VB]      |OH   |M     |
+----------------------+------------------+-----+------+
'''

#6. Filter Based on List Values
li=["OH","CA","DE"]
df.filter(df.state.isin(li)).show(truncate=False)
'''
+----------------------+------------------+-----+------+
|name                  |languages         |state|gender|
+----------------------+------------------+-----+------+
|{James, , Smith}      |[Java, Scala, C++]|OH   |M     |
|{Julia, , Williams}   |[CSharp, VB]      |OH   |F     |
|{Mike, Mary, Williams}|[Python, VB]      |OH   |M     |
+----------------------+------------------+-----+------+
'''

#7. Filter Based on Starts With, Ends With, Contains
df.filter(df.state.startswith("N")).show(truncate=False)
df.filter(df.state.endswith("O")).show(truncate=False)
df.filter(df.state.contains("O")).show(truncate=False)
```

```python
#PySpark Distinct to Drop Duplicate Rows
#Consider below example

# Prepare Data
data = [("James", "Sales", 3000), \
        ("Michael", "Sales", 4600), \
        ("Robert", "Sales", 4100), \
        ("Maria", "Finance", 3000), \
        ("James", "Sales", 3000), \
        ("Scott", "Finance", 3300), \
        ("Jen", "Finance", 3900), \
        ("Jeff", "Marketing", 3000), \
        ("Kumar", "Marketing", 2000), \
        ("Saif", "Sales", 4100) \
]
# Create DataFrame
columns=["employee_name","department","salary"]

df = spark.createDataFrame(data,columns)
df.show(truncate=False)
'''
+-------------+----------+------+
|employee_name|department|salary|
+-------------+----------+------+
|James        |Sales     |3000  |
|Michael      |Sales     |4600  |
|Robert       |Sales     |4100  |
|Maria        |Finance   |3000  |
|James        |Sales     |3000  |
|Scott        |Finance   |3300  |
|Jen          |Finance   |3900  |
|Jeff         |Marketing |3000  |
|Kumar        |Marketing |2000  |
|Saif         |Sales     |4100  |
+-------------+----------+------+
'''

originalDFCount = df.count()
print(f'Original DF count: {originalDFCount}')

#1:Get Distinct Rows (By Comparing All Columns)
DistinctCount = df.distinct().count()
print(f"Removed duplicates of count : {DistinctCount}")
'''
Original DF count: 10
Removed duplicates of count : 9
'''
df2 = df.dropDuplicates()
print(f"Distinct count of post dropDuplicates: {df2.count()}")
'''
Distinct count of post dropDuplicates: 9
'''
```

```python
#PySpark logical Join Types | Join Two DataFrames
'''
Inner Join: Returns only the rows with matching keys in both DataFrames.
Left Join: Returns all rows from the left DataFrame and matching rows from the right DataFrame.
Right Join: Returns all rows from the right DataFrame and matching rows from the left DataFrame.
Full Outer Join: Returns all rows from both DataFrames, including matching and non-matching rows.
Left Semi Join: Returns all rows from the left DataFrame where there is a match in the right DataFrame.
Left Anti Join: Returns all rows from the left DataFrame where there is no match in the right DataFrame.
'''

emp = [(1,"Smith",-1,"2018","10","M",3000), \
       (2,"Rose",1,"2010","20","M",4000), \
       (3,"Williams",1,"2010","10","M",1000), \
       (4,"Jones",2,"2005","10","F",2000), \
       (5,"Brown",2,"2010","40","",-1), \
       (6,"Brown",2,"2010","50","",-1) \
]

empColumns = ["emp_id","name","superior_emp_id","year_joined", "emp_dept_id","gender","salary"]

dept = [("Finance",10), \
    ("Marketing",20), \
    ("Sales",30), \
    ("IT",40) \
]
deptColumns = ["dept_name","dept_id"]

empDF = spark.createDataFrame(emp,empColumns)
deptDF = spark.createDataFrame(dept,deptColumns)

#Inner Join - Returns rows which are matching from both the dataframe
empDF.alias("e").join(deptDF.alias("d"),col("e.emp_dept_id") == col("d.dept_id"),"inner").show(truncate=False)
"""OR"""
empDF.join(deptDF,empDF.emp_dept_id == deptDF.dept_id,"inner").show(truncate=False)

#Left Join - Return all rows from left dataframe and only matching data from right dataframe
empDF.alias("e").join(deptDF.alias("d"),col("e.emp_dept_id") == col("d.dept_id"),"left").show(truncate=False)
"""OR"""
empDF.join(deptDF,empDF.emp_dept_id == deptDF.dept_id,"left").show(truncate=False)
"""OR"""
empDF.alias("e").join(deptDF.alias("d"),col("e.emp_dept_id") == col("d.dept_id"),"leftouter").show(truncate=False)
"""OR"""
empDF.join(deptDF,empDF.emp_dept_id == deptDF.dept_id,"leftouter").show(truncate=False)
# Right Join - Return all rows from right dataframe and only matching data from left dataframe
empDF.alias("e").join(deptDF.alias("d"),col("e.emp_dept_id") == col("d.dept_id"),"right").show(truncate=False)
empDF.alias("e").join(deptDF.alias("d"),col("e.emp_dept_id") == col("d.dept_id"),"rightouter").show(truncate=False)
empDF.join(deptDF,empDF.emp_dept_id == deptDF.dept_id,"right").show(truncate=False)
# Left Semi  Join - Return all rows which are matching with left dataframe only
empDF.alias("e").join(deptDF.alias("d"),col("e.emp_dept_id") == col("d.dept_id"),"leftsemi").show(truncate=False)
empDF.join(deptDF,empDF.emp_dept_id == deptDF.dept_id,"leftsemi").show(truncate=False)
#Left anti join - Return all rows from left dataframe which are not matched with right dataframe
empDF.alias("e").join(deptDF.alias("d"),col("e.emp_dept_id") == col("d.dept_id"),"leftanti").show(truncate=False)
empDF.join(deptDF,empDF.emp_dept_id == deptDF.dept_id,"leftanti").show(truncate=False)
#7: Self join
from pyspark.sql.functions import col
empDF.alias("emp1").join(empDF.alias("emp2"), \
    col("emp1.superior_emp_id") == col("emp2.emp_id"),"inner") \
    .select(col("emp1.emp_id"),col("emp1.name"), \
      col("emp2.emp_id").alias("superior_emp_id"), \
      col("emp2.name").alias("superior_emp_name")) \
   .show(truncate=False)
```

```python
# PySpark MapType - used to represent map key-value pair similar to python Dictionary (Dict)
#from pyspark.sql.types import StringType, MapType
#mapCol = MapType(StringType(),StringType(),False)

'''
Example:
Create MapType() from StructType
from pyspark.sql.types    import StructField, StructType, StringType, MapType
schema = StructType([
   			 StructField('name', StringType(), True),
    			StructField('properties', MapType(StringType(),StringType()),True)
                                   ])
'''
from pyspark.sql.types import StructType,StructField,StringType,MapType
dataDictionary = [
                  ('James',{'hair':'black','eye':'brown'}),
                  ('Michael',{'hair':'brown','eye':None}),
                  ('Robert',{'hair':'red','eye':'black'}),
                  ('Washington',{'hair':'grey','eye':'grey'}),
                  ('Jefferson',{'hair':'brown','eye':''})
]

schema = StructType([
            StructField("Name",StringType(),True),
            StructField("properties",MapType(StringType(),StringType()),True)

])

df = spark.createDataFrame(dataDictionary,schema)
df.printSchema()
'''
root
 |-- Name: string (nullable = true)
 |-- properties: map (nullable = true)
 |    |-- key: string
 |    |-- value: string (valueContainsNull = true)
 '''
df.show(truncate=False)
'''
+----------+-----------------------------+
|Name      |properties                   |
+----------+-----------------------------+
|James     |{eye -> brown, hair -> black}|
|Michael   |{eye -> NULL, hair -> brown} |
|Robert    |{eye -> black, hair -> red}  |
|Washington|{eye -> grey, hair -> grey}  |
|Jefferson |{eye -> , hair -> brown}     |
+----------+-----------------------------+
'''

#Method - 1:Access PySpark MapType() element
df.rdd.map(lambda x: (x.Name,x.properties["hair"],x.properties["eye"])).toDF(["name","hair","eye"]).show(truncate=False)
'''
+----------+-----+-----+
|name      |hair |eye  |
+----------+-----+-----+
|James     |black|brown|
|Michael   |brown|NULL |
|Robert    |red  |black|
|Washington|grey |grey |
|Jefferson |brown|     |
+----------+-----+-----+
'''

#Method - 2:Using getItem() method by passing key as argument and returns value
df.withColumn("hair",df.properties.getItem("hair")) \
  .withColumn("eye",df.properties.getItem("eye")) \
  .drop("properties") \
  .show(truncate=False)

'''
+----------+-----+-----+
|Name      |hair |eye  |
+----------+-----+-----+
|James     |black|brown|
|Michael   |brown|NULL |
|Robert    |red  |black|
|Washington|grey |grey |
|Jefferson |brown|     |
+----------+-----+-----+
'''
```

```python
#PySpark JSON Functions with Examples:
'''
1.from_json() - Converts JSON string into Struct type or Map type.
2.to_json() - Converts MapType or Struct type to JSON string.
3.json_tuple() - Extract the Data from JSON and create them as a new columns.
4.get_json_object() - Extracts JSON element from a JSON string based on json path specified.
5.schema_of_json() - Create schema string from JSON string
'''
#Consider below example
jsonString="""{"Zipcode":704,"ZipCodeType":"STANDARD","City":"PARC PARQUE","State":"PR"}"""
df = spark.createDataFrame([(1,jsonString)],["id","value"])
df.show(truncate=False)
'''
+---+--------------------------------------------------------------------------+
|id |value                                                                     |
+---+--------------------------------------------------------------------------+
|1  |{"Zipcode":704,"ZipCodeType":"STANDARD","City":"PARC PARQUE","State":"PR"}|
+---+--------------------------------------------------------------------------+
'''

#from_json()- used to convert JSON string into Struct type or Map type.
from pyspark.sql.functions import from_json
from pyspark.sql.types import StructType,StructField,MapType,StringType

df.withColumn("value",from_json(df.value,MapType(StringType(),StringType()))).show(truncate=False)
'''
+---+---------------------------------------------------------------------------+
|id |value                                                                      |
+---+---------------------------------------------------------------------------+
|1  |{Zipcode -> 704, ZipCodeType -> STANDARD, City -> PARC PARQUE, State -> PR}|
+---+---------------------------------------------------------------------------+
'''
```

```python
#to_json()-  used to convert DataFrame columns MapType or Struct type to JSON string.
from pyspark.sql.functions import from_json,to_json
from pyspark.sql.types import MapType
df2 = df.withColumn("value",from_json(col("value"),MapType(StringType(),StringType())))
df2.withColumn("value",to_json(col("value"))).show(truncate=False)
'''
+---+----------------------------------------------------------------------------+
|id |value                                                                       |
+---+----------------------------------------------------------------------------+
|1  |{"Zipcode":"704","ZipCodeType":"STANDARD","City":"PARC PARQUE","State":"PR"}|
+---+----------------------------------------------------------------------------+
'''
```

```python
#json_tuple() - used the query or extract the elements from JSON column and create the result as a new columns.
from pyspark.sql.functions import json_tuple
df.select(col("id"),json_tuple(col("value"),"Zipcode","ZipCodeType","City","State"))\
  .toDF("id","Zipcode","ZipCodeType","City","state") \
  .show(truncate=False)

'''
+---+-------+-----------+-----------+-----+
|id |Zipcode|ZipCodeType|City       |state|
+---+-------+-----------+-----------+-----+
|1  |704    |STANDARD   |PARC PARQUE|PR   |
+---+-------+-----------+-----------+-----+
'''
```

```python
#get_json_object() - used to extract the JSON string based on path from the JSON column.
from pyspark.sql.functions import get_json_object
df.select(col("id"),get_json_object(col("value"), "$.ZipCodeType").alias("ZipCodeType")).show(truncate=False)
'''
+---+-----------+
|id |ZipCodeType|
+---+-----------+
|1  |STANDARD   |
+---+-----------+
'''

#schema_of_json() - Use schema_of_json() to create schema string from JSON string column.
from pyspark.sql.functions import schema_of_json,lit
schemaStr=spark.range(1) \
    .select(schema_of_json(lit("""{"Zipcode":704,"ZipCodeType":"STANDARD","City":"PARC PARQUE","State":"PR"}"""))) \
    .collect()[0][0]
print(schemaStr)
'''
STRUCT<City: STRING, State: STRING, ZipCodeType: STRING, Zipcode: BIGINT>
'''
```

```python
#Pyspark-split():
#To convert a string column (StringType) to an array column (ArrayType) in PySpark,
#you can use the split() function from the pyspark.sql.functions module.
#pyspark.sql.functions.split(str, pattern, limit=-1)
#Consider below example
data = [("James, A, Smith","2018","M",3000),
            ("Michael, Rose, Jones","2010","M",4000),
            ("Robert,K,Williams","2010","M",4000),
            ("Maria,Anne,Jones","2005","F",4000),
            ("Jen,Mary,Brown","2010","",-1)
]
columns=["name","dob_year","gender","salary"]
df = spark.createDataFrame(data,columns)
df.show(truncate=False)
'''
+--------------------+--------+------+------+
|name                |dob_year|gender|salary|
+--------------------+--------+------+------+
|James, A, Smith     |2018    |M     |3000  |
|Michael, Rose, Jones|2010    |M     |4000  |
|Robert,K,Williams   |2010    |M     |4000  |
|Maria,Anne,Jones    |2005    |F     |4000  |
|Jen,Mary,Brown      |2010    |      |-1    |
+--------------------+--------+------+------+
'''
#PySpark Convert String to Array Column
df.select(split("name",",").alias("namedArray")).show(truncate=False)
'''
+------------------------+
|namedArray              |
+------------------------+
|[James,  A,  Smith]     |
|[Michael,  Rose,  Jones]|
|[Robert, K, Williams]   |
|[Maria, Anne, Jones]    |
|[Jen, Mary, Brown]      |
+------------------------+
'''
#Convert String to Array Column using SQL Query
df.createOrReplaceTempView("emp")
spark.sql("""
              select split(name,",") as name_array
              from emp
""").show(truncate=False)
'''
+------------------------+
|name_array              |
+------------------------+
|[James,  A,  Smith]     |
|[Michael,  Rose,  Jones]|
|[Robert, K, Williams]   |
|[Maria, Anne, Jones]    |
|[Jen, Mary, Brown]      |
+------------------------+
'''
```

```python
'''
Overlay() function: overlay() in PySpark replaces a portion of a string column with another string,
                    starting at a specified 1-based index position and for a given length.
Syntax: pyspark.sql.functions.overlay(src, replace, pos, len=-1)

src: The source string column or string literal.

replace: The string/column to insert into the source string.

pos: The starting position (1-based index) where replacement begins.

len: (Optional) The number of characters to remove/replace from src.
                If omitted or set to -1, it defaults to the length of the replace string.
'''
from pyspark.sql.functions import overlay,col,lit
df = spark.createDataFrame([("Hello World",)], ["text"])
df.show(truncate=False)
'''
+-----------+
|text       |
+-----------+
|Hello World|
+-----------+
'''
# Replace starting at position 7 for length equal to "PySpark" (7 chars)
df.withColumn("result",overlay(col("text"),lit("PySpark"),7)).show(truncate=False)
'''
+-----------+-------------+
|text       |result       |
+-----------+-------------+
|Hello World|Hello PySpark|
+-----------+-------------+
'''
# Inserting Without Removing Characters (len = 0):
df.select(overlay(col("text"),lit("PySpark "),7,0).alias("text")).show(truncate=False)

+------------------+
|text              |
+------------------+
|Hello PySpark World|
+------------------+
```

```python
# Replacing a Fixed Length (len > 0):
# Replace 5 characters starting at position 7 with "Spark"
df.withColumn("result", overlay(col("text"), lit("Spark"), 7, 5)).show(truncate=False)
'''
+-----------+-----------+
|text       |result     |
+-----------+-----------+
|Hello World|Hello Spark|
+-----------+-----------+
'''
```

```python
# PySpark – Difference between two dates (days, months, years):
data = [("1","2019-07-01"),("2","2019-06-24"),("3","2019-08-24")]
df=spark.createDataFrame(data=data,schema=["id","date"])

df.show(truncate=False)
'''
+---+----------+
|id |date      |
+---+----------+
|1  |2019-07-01|
|2  |2019-06-24|
|3  |2019-08-24|
+---+----------+
'''
from pyspark.sql.functions import current_date,datediff
df.select("id","date",current_date().alias("current_date"),datediff(current_date(),"date").alias("dateInBetween")).show(truncate=False)
'''
+---+----------+------------+-------------+
|id |date      |current_date|dateInBetween|
+---+----------+------------+-------------+
|1  |2019-07-01|2026-09-09  |2627         |
|2  |2019-06-24|2026-09-09  |2634         |
|3  |2019-08-24|2026-09-09  |2573         |
+---+----------+------------+-------------+
'''

#months_between() Function:
from pyspark.sql.functions import months_between,round
df.select("id","date",current_date().alias("current_date"),round(months_between(current_date(),"date"),2).alias("months_Between")).show(truncate=False)
'''
+---+----------+------------+--------------+
|id |date      |current_date|months_Between|
+---+----------+------------+--------------+
|1  |2019-07-01|2026-09-09  |86.26         |
|2  |2019-06-24|2026-09-09  |86.52         |
|3  |2019-08-24|2026-09-09  |84.52         |
+---+----------+------------+--------------+
'''
#yearsBetween:
df.select("id","date",current_date().alias("current_date"),round(months_between(current_date(),"date")/lit(12),2).alias("years_between")).show(truncate=False)

'''
+---+----------+------------+-------------+
|id |date      |current_date|years_between|
+---+----------+------------+-------------+
|1  |2019-07-01|2026-09-09  |7.19         |
|2  |2019-06-24|2026-09-09  |7.21         |
|3  |2019-08-24|2026-09-09  |7.04         |
+---+----------+------------+-------------+
'''
```

```python
# PySpark SQL – Working with Unix Time | Timestamp:
#unix_timestamp() – Converts Date and Timestamp Column to Unix Time:
'''
What is Unix Time?
In PySpark, Unix time (also known as POSIX time or Epoch time) represents the total number of seconds that have elapsed
since January 1, 1970, 00:00:00 UTC.
'''

inputData = [("2019-07-01 12:01:19",
            "07-01-2019 12:01:19",
            "07-01-2019")]
columns=["timestamp_1","timestamp_2","timestamp_3"]
df=spark.createDataFrame(
        data = inputData,
        schema = columns)
df.printSchema()
df.show(truncate=False)

'''
root
 |-- timestamp_1: string (nullable = true)
 |-- timestamp_2: string (nullable = true)
 |-- timestamp_3: string (nullable = true)

+-------------------+-------------------+-----------+
|timestamp_1        |timestamp_2        |timestamp_3|
+-------------------+-------------------+-----------+
|2019-07-01 12:01:19|07-01-2019 12:01:19|07-01-2019 |
+-------------------+-------------------+-----------+
'''
from pyspark.sql.functions import unix_timestamp,col
df.select(
      unix_timestamp(col("timestamp_1")).alias("timestamp_1"),
      unix_timestamp(col("timestamp_2"),"MM-dd-yyyy HH:mm:ss").alias("timestamp_2"),
      unix_timestamp(col("timestamp_3"),"MM-dd-yyyy").alias("timestamp_3"),
      unix_timestamp().alias("timestamp_4")
).show(truncate=False)

'''
+-----------+-----------+-----------+-----------+
|timestamp_1|timestamp_2|timestamp_3|timestamp_4|
+-----------+-----------+-----------+-----------+
|1561982479 |1561982479 |1561939200 |1788945944 |
+-----------+-----------+-----------+-----------+
'''

#from_unixtime() – Converts Unix Time Seconds to Date and Timestamp
from pyspark.sql.functions import from_unixtime,col
df.select(
      from_unixtime(col("timestamp_1")).alias("timestamp_1"),
      from_unixtime(col("timestamp_2"),"MM-dd-yyyy HH:mm:ss").alias("timestamp_2"),
      from_unixtime(col("timestamp_3"),"MM-dd-yyyy").alias("timestamp_3"),
      from_unixtime(col("timestamp_4")).alias("timestamp_4")

).show(truncate=False)

```

```python
#PySpark : explode nested array into rows
#In PySpark, the explode() function creates a new row for each element in a given array or map column.

arrayArrayData = [
  ("James",[["Java","Scala","C++"],["Spark","Java"]]),
  ("Michael",[["Spark","Java","C++"],["Spark","Java"]]),
  ("Robert",[["CSharp","VB"],["Spark","Python"]])
]

df = spark.createDataFrame(data=arrayArrayData, schema = ['name','subjects'])
df.printSchema()
df.show(truncate=False)
'''
root
 |-- name: string (nullable = true)
 |-- subjects: array (nullable = true)
 |    |-- element: array (containsNull = true)
 |    |    |-- element: string (containsNull = true)

+-------+-----------------------------------+
|name   |subjects                           |
+-------+-----------------------------------+
|James  |[[Java, Scala, C++], [Spark, Java]]|
|Michael|[[Spark, Java, C++], [Spark, Java]]|
|Robert |[[CSharp, VB], [Spark, Python]]    |
+-------+-----------------------------------+
'''

from pyspark.sql.functions import explode
df.select("name",explode("subjects")).show(truncate=False)
'''
+-------+------------------+
|name   |col               |
+-------+------------------+
|James  |[Java, Scala, C++]|
|James  |[Spark, Java]     |
|Michael|[Spark, Java, C++]|
|Michael|[Spark, Java]     |
|Robert |[CSharp, VB]      |
|Robert |[Spark, Python]   |
+-------+------------------+
'''
#If you want to flatten the arrays, use flatten function which converts array of array columns to a single array on DataFrame.
from pyspark.sql.functions import flatten
df.select("name",flatten("subjects")).show(truncate=False)
'''
+-------+-------------------------------+
|name   |flatten(subjects)              |
+-------+-------------------------------+
|James  |[Java, Scala, C++, Spark, Java]|
|Michael|[Spark, Java, C++, Spark, Java]|
|Robert |[CSharp, VB, Spark, Python]    |
+-------+-------------------------------+
'''

#explode() - PySpark explode array or map column to rows.
arrayData = [
                        ('James',['Java','Scala'],{'hair':'black','eye':'brown'}),
                        ('Michael',['Spark','Java',None],{'hair':'brown','eye':None}),
                        ('Robert',['CSharp',''],{'hair':'red','eye':''}),
                        ('Washington',None,None),
                        ('Jefferson',['1','2'],{})
]
df = spark.createDataFrame(data=arrayData, schema = ['name','knownLanguages','properties'])
df.show(truncate=False)
'''
+----------+-------------------+-----------------------------+
|name      |knownLanguages     |properties                   |
+----------+-------------------+-----------------------------+
|James     |[Java, Scala]      |{eye -> brown, hair -> black}|
|Michael   |[Spark, Java, NULL]|{eye -> NULL, hair -> brown} |
|Robert    |[CSharp, ]         |{eye -> , hair -> red}       |
|Washington|NULL               |NULL                         |
|Jefferson |[1, 2]             |{}                           |
+----------+-------------------+-----------------------------+
'''

df.select(explode("knownLanguages"),explode("properties")).show(truncate=False)
'''
+------+----+-----+
|col   |key |value|
+------+----+-----+
|Java  |eye |brown|
|Java  |hair|black|
|Scala |eye |brown|
|Scala |hair|black|
|Spark |eye |NULL |
|Spark |hair|brown|
|Java  |eye |NULL |
|Java  |hair|brown|
|NULL  |eye |NULL |
|NULL  |hair|brown|
|CSharp|eye |     |
|CSharp|hair|red  |
|      |eye |     |
|      |hair|red  |
+------+----+-----+
'''
```

```python
# collect_list() : returns all values from an input column with duplicates.
from pyspark.sql.functions import collect_list
data = [
    ("Alice", "Apple"),
    ("Alice", "Banana"),
    ("Alice", "Apple"), # Duplicate entry preserved
    ("Bob", "Orange"),
    ("Bob", "Banana")
]

df = spark.createDataFrame(data, ["name", "fruit"])
df.show(truncate=False)
'''
+-----+------+
|name |fruit |
+-----+------+
|Alice|Apple |
|Alice|Banana|
|Alice|Apple |
|Bob  |Orange|
|Bob  |Banana|
+-----+------+
'''

df.select(collect_list("fruit")).show(truncate=False)
'''
+--------------------------------------+
|collect_list(fruit)                   |
+--------------------------------------+
|[Apple, Banana, Apple, Orange, Banana]|
+--------------------------------------+
'''

df.groupBy("name").agg(collect_list("fruit")).show(truncate=False)
'''
+-----+----------------------+
|name |collect_list(fruit)   |
+-----+----------------------+
|Alice|[Apple, Banana, Apple]|
|Bob  |[Orange, Banana]      |
+-----+----------------------+
'''

#collect_set() function returns all values from an input column with duplicate values eliminated.
from pyspark.sql.functions import collect_set
df.groupBy("name").agg(collect_set("fruit")).show(truncate=False)
'''
+-----+------------------+
|name |collect_set(fruit)|
+-----+------------------+
|Alice|[Apple, Banana]   |
|Bob  |[Orange, Banana]  |
+-----+------------------+
'''

```

```python
# create_map() function: used to Convert DataFrame Columns to MapType
from pyspark.sql.types import *
data = [("36636","Finance",3000,"USA"),
        ("40288","Finance",5000,"IND"),
        ("42114","Sales",3900,"USA"),
        ("39192","Marketing",2500,"CAN"),
        ("34534","Sales",6500,"USA")]

schema = StructType([
     StructField('id', StringType(), True),
     StructField('dept', StringType(), True),
     StructField('salary', IntegerType(), True),
     StructField('location', StringType(), True)
     ])

df = spark.createDataFrame(data,schema)
df.show(truncate=False)
'''
+-----+---------+------+--------+
|id   |dept     |salary|location|
+-----+---------+------+--------+
|36636|Finance  |3000  |USA     |
|40288|Finance  |5000  |IND     |
|42114|Sales    |3900  |USA     |
|39192|Marketing|2500  |CAN     |
|34534|Sales    |6500  |USA     |
+-----+---------+------+--------+
'''

from pyspark.sql.functions import create_map,lit
df.withColumn("propertiesMap", create_map(
    lit("salary"), col("salary").cast("string"),
    lit("location"), col("location").cast("string")
)).drop("salary", "location").show(truncate=False)

'''
+-----+---------+---------------------------------+
|id   |dept     |propertiesMap                    |
+-----+---------+---------------------------------+
|36636|Finance  |{salary -> 3000, location -> USA}|
|40288|Finance  |{salary -> 5000, location -> IND}|
|42114|Sales    |{salary -> 3900, location -> USA}|
|39192|Marketing|{salary -> 2500, location -> CAN}|
|34534|Sales    |{salary -> 6500, location -> USA}|
+-----+---------+---------------------------------+
'''
```

```python
#countDIstinct()-To get the count distinct
data = [("James", "Sales", 3000),
    ("Michael", "Sales", 4600),
    ("Robert", "Sales", 4100),
    ("Maria", "Finance", 3000),
    ("James", "Sales", 3000),
    ("Scott", "Finance", 3300),
    ("Jen", "Finance", 3900),
    ("Jeff", "Marketing", 3000),
    ("Kumar", "Marketing", 2000),
    ("Saif", "Sales", 4100)
  ]
columns = ["Name","Dept","Salary"]
df = spark.createDataFrame(data=data,schema=columns)
df.show(truncate=False)
'''
+-------+---------+------+
|Name   |Dept     |Salary|
+-------+---------+------+
|James  |Sales    |3000  |
|Michael|Sales    |4600  |
|Robert |Sales    |4100  |
|Maria  |Finance  |3000  |
|James  |Sales    |3000  |
|Scott  |Finance  |3300  |
|Jen    |Finance  |3900  |
|Jeff   |Marketing|3000  |
|Kumar  |Marketing|2000  |
|Saif   |Sales    |4100  |
+-------+---------+------+
'''
# Applying distinct() and count() together:
df.select("Name", "Dept", "Salary").distinct().count() #9

#Applying countDistinct() function
from pyspark.sql.functions import countDistinct
df.select(
    countDistinct("Name").alias("unique_names"),
    countDistinct("Dept").alias("unique_depts"),
    countDistinct("Salary").alias("unique_salaries")
).show(truncate=False)
'''
+------------+------------+---------------+
|unique_names|unique_depts|unique_salaries|
+------------+------------+---------------+
|9           |3           |6              |
+------------+------------+---------------+
'''

#Applying collect() after countDistinct():
print("Distinct Count of Department & Salary: ",df.select(countDistinct("Name","Dept")).collect()[0][0])
# Distinct Count of Department & Salary:  9

#If want to display distinct & count all records
print("Distinct Count: ",df.distinct().count())
#Distinct Count:  9
```

```python
#Pyspark lit() - Add Literal or Constant to Dataframe:

data = [("111",50000),("222",60000),("333",40000)]
columns= ["EmpId","Salary"]
df = spark.createDataFrame(data = data, schema = columns)
df.show(truncate=False)
'''
+-----+------+
|EmpId|Salary|
+-----+------+
|111  |50000 |
|222  |60000 |
|333  |40000 |
+-----+------+
'''
# lit() Function to Add Constant Column:
from pyspark.sql.functions import lit
df.select("EmpId","Salary",lit(1)).show(truncate=False)
'''
+-----+------+---+
|EmpId|Salary|1  |
+-----+------+---+
|111  |50000 |1  |
|222  |60000 |1  |
|333  |40000 |1  |
+-----+------+---+
'''
from pyspark.sql.functions import when,col
df.withColumn("lit_value2", when((col("Salary") >=40000) & (col("Salary") <=   50000),lit("100")) \
                           .otherwise(lit("200"))).show(truncate=False)
'''
+-----+------+----------+
|EmpId|Salary|lit_value2|
+-----+------+----------+
|111  |50000 |100       |
|222  |60000 |200       |
|333  |40000 |100       |
+-----+------+----------+
'''

----------------------------------------------------------------End Of PySpark-------------------------------------------------------------
```
